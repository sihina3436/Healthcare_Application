```
import cv2
from google.colab.patches import cv2_imshow

img = cv2.imread('image_1.jpg')
grayscale_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
cv2_imshow(grayscale_img)
```

```
import cv2
from google.colab.patches import cv2_imshow

img = cv2.imread('image_1.jpg')
resize_img = cv2.resize(img,(600,400))

cv2_imshow(resize_img)
```

```
import cv2
from google.colab.patches import cv2_imshow

img = cv2.imread('image_1.jpg')
rotate_img = cv2.rotate(img, cv2.ROTATE_180)

cv2_imshow(rotate_img)
```

```
import cv2
from google.colab.patches import cv2_imshow

img = cv2.imread('image_1.jpg')
blur_img = cv2.GaussianBlur(img, (21,21), 0)

cv2_imshow(blur_img)
```
